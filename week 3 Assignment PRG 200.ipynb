```python
balance = 50000
daily_withdrawn = 10000
amount = int(input("Enter withdrawal amount: "))

if amount % 500 != 0:
    print("Invalid amount. Must be a multiple of NPR 500.")

elif daily_withdrawn + amount > 50000:
    print("Daily withdrawal limit reached.")

elif amount > balance:
    print("Insufficient balance.")

else:
    balance -= amount
    daily_withdrawn += amount
    print("Withdrawal successful.")
    print("Your current balance after withdrawal: NPR", balance)
```

```python
purchase = int(input("Enter total purchase amount: "))
is_member = input("Are you a loyalty member? (yes/no): ")

discount = 0

if purchase < 1000:
    discount = 0

elif purchase <= 4999:
    discount = 0.05

elif purchase <= 14999:
    discount = 0.10

else:
    discount = 0.20

discounted_amount = purchase - (purchase * discount)

# extra loyalty discount
if is_member.lower() == "yes":
    discounted_amount -= discounted_amount * 0.05

print("Final payable amount:", discounted_amount)
```

```python
inventory = [
    {"item": "Rice", "stock": 5, "threshold": 10},
    {"item": "Eggs", "stock": 24, "threshold": 12},
    {"item": "Milk", "stock": 3, "threshold": 6},
    {"item": "Bread", "stock": 8, "threshold": 5},
    {"item": "Chicken", "stock": 0, "threshold": 4},
    {"item": "Cooking Oil", "stock": 2, "threshold": 3},
]

count = 0

for item in inventory:
    if item["stock"] < item["threshold"]:
        print("Restock alert:", item["item"])
        count += 1

print("Total items to restock:", count)
```

```python
passwords = ["hello", "Hello123", "H3ll0@World", "12345678", "MyP@ss!"]
special_chars = "!@#$%^&*"

for pw in passwords:
    errors = []

    if len(pw) < 8:
        errors.append("At least 8 characters")

    if not any(c.isupper() for c in pw):
        errors.append("At least one uppercase letter")

    if not any(c.islower() for c in pw):
        errors.append("At least one lowercase letter")

    if not any(c.isdigit() for c in pw):
        errors.append("At least one digit")

    if not any(c in special_chars for c in pw):
        errors.append("At least one special character")

    if len(errors) == 0:
        print(pw, "-> Strong Password")

    else:
        print(pw, "-> Weak Password | Missing:", ", ".join(errors))
```

```python
trips = [
    {"distance": 1.5, "hour": 14},
    {"distance": 5.0, "hour": 22},
    {"distance": 12.0, "hour": 3},
    {"distance": 8.5, "hour": 10},
    {"distance": 2.0, "hour": 23},
]

for trip in trips:
    distance = trip["distance"]
    hour = trip["hour"]

    fare = 150  # base fare

    if distance > 2:
        if distance <= 10:
            fare += (distance - 2) * 35

        else:
            fare += 8 * 35
            fare += (distance - 10) * 28

    # night surcharge
    if hour >= 22 or hour < 5:
        fare += fare * 0.10

    print("Trip:", trip, "=> Fare:", round(fare, 2))
```
